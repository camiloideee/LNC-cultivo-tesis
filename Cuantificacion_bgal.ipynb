```python
import matplotlib.pyplot as plt
import numpy as np

# Datos
grupos = ["DMSO", "ABT", "DMSO", "ABT"]
valores = [22.57, 33.19, 67.59, 50.52]

# Posiciones en x: dos barras para Tubo Neural y dos para Notocorda
x = np.array([0, 1, 3, 4])

# Colores por condición
color_dmso = "#8ad4c1"
color_abt = "#df78bd"

colores = [
    color_dmso,  # Tubo Neural DMSO
    color_abt,   # Tubo Neural ABT
    color_dmso,  # Notocorda DMSO
    color_abt    # Notocorda ABT
]

fig, ax = plt.subplots(figsize=(6, 5))

# Barras
ax.bar(
    x,
    valores,
    color=colores,
    edgecolor="black",
    linewidth=0.8,
    width=0.75
)

# Puntitos marcando el valor de cada barra
ax.scatter(
    x,
    valores,
    color="black",
    s=35,
    zorder=3
)

# Etiquetas de condiciones
ax.set_xticks(x)
ax.set_xticklabels(grupos, fontsize=11)

# Etiquetas de estructura debajo de cada grupo
ax.text(0.5, -8, "Tubo Neural", ha="center", va="top", fontsize=12)
ax.text(3.5, -8, "Notocorda", ha="center", va="top", fontsize=12)

# Eje y
ax.set_ylabel(r"Promedio actividad SA-$\beta$-gal / $\mu m^2$", fontsize=12)
ax.set_ylim(0, 95)

# Estética
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```
